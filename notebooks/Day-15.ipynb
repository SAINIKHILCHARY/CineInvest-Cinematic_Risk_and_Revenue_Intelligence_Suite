```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv(
    r"C:\Users\91934\OneDrive\Desktop\DA\ZDIP\data\cleaned\cleaned_movies.csv"
)
```

#### Correlation Matrix Analysis

```python
columns = ["Revenue", "Genre_Count", "Production_Company_Count"]

correlation_matrix = df[columns].corr()

print(correlation_matrix)
```

#### Heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.show()
```

#### Variance Analysis

```python
print(
    "Variance of Cast_Gender_Ratio:",
    df["Cast_Gender_Ratio"].var()
)
```

```python
variance_analysis = (
    df.groupby("Primary_Production_Company")["Revenue"]
      .var()
      .sort_values(ascending=False)
)

print(variance_analysis.head(10))
```

```python
company_count = (
    df.groupby("Primary_Production_Company")
      .size()
      .sort_values(ascending=False)
)

print(company_count.head(10))
```

#### Statistical Validation of the Main Hypothesis

Hypothesis:

- Movies with more genres and more production companies tend to generate higher revenue.

Validation steps:

- Check the correlation between Revenue and Genre_Count.
- Check the correlation between Revenue and Production_Company_Count.
- Compare variance across production companies.
- Interpret whether the relationships are strong, moderate, or weak.

#### Limitations
- Correlation does not imply causation.
- Extreme revenue values can influence the analysis.
- Missing values may affect the results.
- Some production companies have very few movies.
- External factors such as marketing and audience preferences are not included in the dataset.
